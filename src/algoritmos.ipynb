# Apriori


```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

df = pd.read_csv("../datos/datos_preproc.csv", header=0)

basket = pd.crosstab(df["Transaction"], df["Item"]).gt(0)
basket.head()
```

```python
for s in [0.005, 0.008, 0.01, 0.02, 0.03]:
    fi = apriori(basket, min_support=s, use_colnames=True)
    print(s, len(fi), fi["itemsets"].apply(len).value_counts().sort_index().to_dict())
```

Viendo esto, creemos que la mejor forma de trabajar es con 0.01, ya que tiene 3 tripletas, al igual 0.008, pero con 28 duplas, una cantidad manejable de información. 

```python
fi = apriori(basket, min_support=0.01, use_colnames=True)
fi["length"] = fi["itemsets"].apply(len)

rules = association_rules(fi, num_itemsets=len(basket), metric="lift", min_threshold=1.2)

cols = ["antecedents", "consequents", "support", "confidence", "lift", "leverage", "conviction"]
rules[cols].sort_values("lift", ascending=False).head(15).to_csv("../datos/rules.csv")
```

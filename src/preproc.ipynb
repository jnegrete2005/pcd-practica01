```python
import pandas as pd

df = pd.read_csv("../datos/datos.xls", header=0)
df.head()
```

```python
df = df.groupby(list(df.columns), as_index=False).size().rename(columns={"size": "count"})
df.head()
```

```python
order = list(df["period_day"].unique())
order
```

```python
df["date_time"] = pd.to_datetime(df["date_time"], format="%d-%m-%Y %H:%M")

df["weekday"] = df["weekday_weekend"] == "weekday"
df = df.drop(columns="weekday_weekend")

df["period_day"] = pd.Categorical(df["period_day"], categories=order, ordered=True).codes
```

```python
df.head(10)
```

```python
df.to_csv("../datos/datos_preproc.csv", sep=",", header=True, index=False)
```
